```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict, Literal
```

```python
class QuadraticState(TypedDict):

    a: int
    b: int
    c: int

    discriminant: float
    expression: str
    roots: str
```

```python
graph = StateGraph(QuadraticState)
```

```python
def show_expression(state: QuadraticState):

    expression = f"{state['a']}x2{state['b']}x{state['c']}"

    return {"expression": expression}
```

```python
def calculate_discriminant(state: QuadraticState):

    discriminant = (state['b']**2) - (4 * state['a'] * state['c'])

    return {"discriminant": discriminant}
```

```python
#   yese bhi handle kar skte hai , par isme conditional workflow nahi bnega , isliye check discrimnant function bnaya

def calculate_roots(state: QuadraticState):

    if state['discriminant'] == 0:
        roots = -state['b'] / (2 * state['a'])

        return {"roots": f"The roots are real and equal {roots}"}
    elif state['discriminant'] > 0:

        root1 = (-state['b'] + (state['discriminant']** 0.5)) / (2 * state['a'])
        root2 = (-state['b'] - (state['discriminant']** 0.5)) / (2 * state['a'])

        return {"roots": f"The roots are real and unequal {root1} & {root2}"}
    else:
        return {"roots": f"There is no real roots exists for this quadratic equation"}
```

```python
def calculate_distinct_roots(state: QuadraticState):
    root1 = (-state['b'] + (state['discriminant']** 0.5)) / (2 * state['a'])
    root2 = (-state['b'] - (state['discriminant']** 0.5)) / (2 * state['a'])

    return {"roots": f"The roots are real and unequal {root1} & {root2}"}

def calculate_same_roots(state: QuadraticState):
    roots = -state['b'] / (2 * state['a'])

    return {"roots": f"The roots are real and equal {roots}"} 

def calculate_no_roots(state: QuadraticState):
    return {"roots": f"There is no real roots exists for this quadratic equation"}
```

```python
def check_discriminant(state: QuadraticState) -> Literal ["calculate_distinct_roots", "calculate_same_roots", "calculate_no_roots"]:
    if state['discriminant'] > 0:
        return "calculate_distinct_roots"
    elif state['discriminant'] == 0:
        return "calculate_same_roots"
    else:
        return "calculate_no_roots"
```

```python
# defining nodes 

graph.add_node("show_expression", show_expression)
graph.add_node("calculate_discriminant", calculate_discriminant)
# graph.add_node("calculate_roots", calculate_roots)
graph.add_node("calculate_distinct_roots", calculate_distinct_roots)
graph.add_node("calculate_same_roots", calculate_same_roots)
graph.add_node("calculate_no_roots", calculate_no_roots)


graph.add_edge(START, "show_expression")
graph.add_edge("show_expression", "calculate_discriminant")

graph.add_conditional_edges("calculate_discriminant", check_discriminant)

graph.add_edge("calculate_same_roots",END)
graph.add_edge("calculate_no_roots", END)
graph.add_edge("calculate_distinct_roots", END)

workflow = graph.compile()
workflow
```

```python
initial_state = {
    "a": 1,
    "b":-4,
    "c":7
}

final_result = workflow.invoke(initial_state)
final_result
```
